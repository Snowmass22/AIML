```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.svm import SVR
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
from sklearn import datasets
```

```python
df=datasets.load_diabetes(as_frame=True).frame
```

```python
df.head()
```

```python
x=df.drop("target",axis=1)
x.head()
```

```python
y=df["target"]
y.head()
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
#scaling the target value
y_scaler=StandardScaler()
y_train_scaled=y_scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
y_test_scaled=y_scaler.fit_transform(y_test.values.reshape(-1,1)).ravel()
```

```python
model=SVR()
model.fit(x_train,y_train_scaled)
```

```python
y_pred_scaled=model.predict(x_test)
```

```python
print("r2 score:",r2_score(y_test_scaled,y_pred_scaled))
```

```python
#hyper parameter using gridsearch cv
from sklearn.model_selection import GridSearchCV
param_grid={
    "C":[1,2,5,10,50,100],
    "kernal":["rbf","linear"],
    "epsilon":[0.01,0.1,0.2,0.3,0.5]
}
```

```python
svr=SVR()
grid_search=GridSearchCV(svr,param_grid,scoring="r2",cv=5)
grid_search.fit(x_train,y_train_scaled)
```
