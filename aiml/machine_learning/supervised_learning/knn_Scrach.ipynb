```python
#knn implementation from scratch (classification)
# knn implementation from scratch (classification)
class KNNClassifier:
    def __init__(self, k=3):
        self.k = k
        
    def _euclidean_dist(self, x1, x2):
        return np.sqrt(np.sum((x1 - x2) ** 2))
        
    def fit(self, x, y):
        # lazy learner
        self.x_train = x
        self.y_train = y
        
    def predict_one(self, x):
        distances = [self._euclidean_dist(x, x_train) for x_train in self.x_train]
        knn_indices = np.argsort(distances)[:self.k]
        knn_classes = [self.y_train[i] for i in knn_indices]
        majority_class = np.argmax(np.bincount(knn_classes))
        return majority_class  # <-- Fixed typo here
        
    def predict(self, x):
        y_pred = [self.predict_one(i) for i in x]  # <-- Fixed underscore and variable masking
        return y_pred
```

```python
import numpy as np

# 1. Define some mock training data (e.g., 4 points with 2 features each)
x_train = np.array([[1, 2], [2, 3], [3, 1], [6, 7]])
y_train = np.array([0, 0, 0, 1])  # Class labels for the points

# 2. Define a test point you want to predict
x_test = np.array([[2, 2]])

# 3. Initialize, fit, and predict
model = KNNClassifier(k=3)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)

print("Predicted class:", y_pred)

```
